```python
# Import Packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
d25 = pd.read_csv('2025Data.csv')
d26 = pd.read_csv('2026Data.csv')

d25 = d25[(d25['Geography'] == 'Total U.S.') & (d25['Period'] <= 32)]
d26 = d26[(d26['Geography'] == 'Total U.S.') & (d26['Period'] <= 32)]

# Revenue
d25['Rev'] = d25['ASP Current Year'] * d25['Total Bulk and Bags']
d26['Rev'] = d26['ASP Current Year'] * d26['Total Bulk and Bags']

# Conventional and Organic
t25 = d25.groupby('Type')[['Total Bulk and Bags', 'Rev']].sum()
t26 = d26.groupby('Type')[['Total Bulk and Bags', 'Rev']].sum()

# Total U.S. 
t25.loc['Total U.S.'] = t25.sum()
t26.loc['Total U.S.'] = t26.sum()

# Table
table = pd.DataFrame()
table['2025 Vol (M)'] = (t25['Total Bulk and Bags'] / 1_000_000).round(2)
table['2026 Vol (M)'] = (t26['Total Bulk and Bags'] / 1_000_000).round(2)
table['Vol YoY %'] = (((t26['Total Bulk and Bags'] - t25['Total Bulk and Bags']) / t25['Total Bulk and Bags']) * 100).round(2)

table['2025 ASP ($)'] = (t25['Rev'] / t25['Total Bulk and Bags']).round(2)
table['2026 ASP ($)'] = (t26['Rev'] / t26['Total Bulk and Bags']).round(2)
table['ASP YoY %'] = (((t26['Rev'] / t26['Total Bulk and Bags'] - t25['Rev'] / t25['Total Bulk and Bags']) / (t25['Rev'] / t25['Total Bulk and Bags'])) * 100).round(2)

table['2025 Rev ($M)'] = (t25['Rev'] / 1_000_000).round(2)
table['2026 Rev ($M)'] = (t26['Rev'] / 1_000_000).round(2)
table['Rev YoY %'] = (((t26['Rev'] - t25['Rev']) / t25['Rev']) * 100).round(2)

print(table)
```
